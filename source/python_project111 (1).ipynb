```python
print("TEST CASE 1: NORMAL INPUT")

import pandas as pd

# Read the uploaded restaurant sales file
test_df = pd.read_csv("restaurant_sales.xls")

# Calculate sales
test_df["Sales"] = test_df["Quantity"] * test_df["Price"]

# Display sales by category
print("\nSales by Category:")
print(test_df.groupby("Category")["Sales"].sum())

# Display sales by date
print("\nSales by Date:")
print(test_df.groupby("Date")["Sales"].sum())

# Find the top-selling item by quantity
print("\nTop Item by Quantity:")
print(test_df.groupby("Item")["Quantity"].sum().idxmax())

# Find the top-selling item by revenue
print("\nTop Item by Revenue:")
print(test_df.groupby("Item")["Sales"].sum().idxmax())

print("\nTest Case 1 completed successfully!")
```

```python
print("TEST CASE 2: BOUNDARY INPUT - ONE ROW")

import pandas as pd

# Create a temporary dataset with only one row
one_row = pd.DataFrame({
    "Date": ["2026-10-01"],
    "Item": ["Dal Makhani"],
    "Category": ["Main"],
    "Quantity": [1],
    "Price": [240]
})

# Calculate sales
one_row["Sales"] = one_row["Quantity"] * one_row["Price"]

print("\nSingle Row Dataset:")
print(one_row)

print("\nSales by Category:")
print(one_row.groupby("Category")["Sales"].sum())

print("\nSales by Date:")
print(one_row.groupby("Date")["Sales"].sum())

print("\nTop Item by Quantity:")
print(one_row.groupby("Item")["Quantity"].sum().idxmax())

print("\nTop Item by Revenue:")
print(one_row.groupby("Item")["Sales"].sum().idxmax())

print("\nTest Case 2 completed successfully!")
```

```python
print("TEST CASE 3: INVALID INPUT")

import pandas as pd

# Part A: Invalid quantity
invalid_data = pd.DataFrame({
    "Date": ["2026-10-01", "2026-10-02"],
    "Item": ["Dal Makhani", "Masala Tea"],
    "Category": ["Main", "Beverage"],
    "Quantity": [1, "invalid"],
    "Price": [240, 40]
})

# Convert invalid quantities to NaN
invalid_data["Quantity"] = pd.to_numeric(
    invalid_data["Quantity"], errors="coerce"
)

# Skip rows with invalid required values
valid_data = invalid_data.dropna(
    subset=["Date", "Item", "Category", "Quantity", "Price"]
).copy()

valid_data["Sales"] = valid_data["Quantity"] * valid_data["Price"]

print("\nValid Rows After Removing Invalid Quantity:")
print(valid_data)

# Part B: Missing CSV file
try:
    missing_df = pd.read_csv("missing_restaurant_sales.csv")
except FileNotFoundError:
    print("\nError handled: CSV file not found.")

print("\nTest Case 3 completed successfully!")
```

```python
print("TEST CASE 4: DIFFERENT SCENARIO")

import pandas as pd

# Create a different temporary dataset
scenario_data = pd.DataFrame({
    "Date": [
        "2026-10-01",
        "2026-10-01",
        "2026-10-02",
        "2026-10-02",
        "2026-10-03"
    ],
    "Item": [
        "Pizza",
        "Burger",
        "Pizza",
        "Pasta",
        "Burger"
    ],
    "Category": [
        "Main",
        "Main",
        "Main",
        "Main",
        "Main"
    ],
    "Quantity": [5, 3, 4, 6, 7],
    "Price": [200, 100, 200, 150, 100]
})

# Calculate sales
scenario_data["Sales"] = (
    scenario_data["Quantity"] * scenario_data["Price"]
)

# Daily sales
print("\nDaily Sales:")
print(scenario_data.groupby("Date")["Sales"].sum())

# Total quantity sold per item
print("\nQuantity Sold by Item:")
print(scenario_data.groupby("Item")["Quantity"].sum())

# Total revenue per item
print("\nRevenue by Item:")
print(scenario_data.groupby("Item")["Sales"].sum())

# Top-selling item by quantity
print("\nTop Item by Quantity:")
print(scenario_data.groupby("Item")["Quantity"].sum().idxmax())

# Top-selling item by revenue
print("\nTop Item by Revenue:")
print(scenario_data.groupby("Item")["Sales"].sum().idxmax())

print("\nTest Case 4 completed successfully!")
```
